```python
import cv2
import numpy as np
from typing import List, Dict
```

```python
# ========== Field Geometry (yards) ==========

FIELD_LENGTH = 100.0   # 0..100 goal line to goal line
FIELD_WIDTH  = 53.3         # NFL width

# Hash marks: 70'9" from sideline → 23.583 yards
HASH_DIST_FROM_SIDELINE = 70.75 / 3.0
HASH_NEAR_Y = HASH_DIST_FROM_SIDELINE
HASH_FAR_Y  = FIELD_WIDTH - HASH_DIST_FROM_SIDELINE

# Numbers approx 12 yards from sideline (centered)
NUM_DIST_FROM_SIDELINE = 12.0
NUM_NEAR_Y = NUM_DIST_FROM_SIDELINE
NUM_FAR_Y  = FIELD_WIDTH - NUM_DIST_FROM_SIDELINE
```

```python
def box_center(bbox):
    x1, y1, x2, y2 = bbox
    return np.array([(x1 + x2) / 2.0, (y1 + y2) / 2.0], dtype=np.float32)
```

```python
def parse_yard_and_side_from_label(label: str):
    """
    Expected formats (adjust to your YOLO class names):
      'hash_40_near'
      'num_30_far'
    Returns: yard (int), side ('near'/'far'), kind ('hash'/'num')
    """
    parts = label.split("_")
    if len(parts) != 3:
        raise ValueError(f"Unexpected class label format: {label}")
    
    kind = parts[0]     # hash or num
    yard = int(parts[1])
    side = parts[2]     # near or far
    return yard, side, kind
```

```python
def build_homography_points(field_dets, frame_bgr, number_model,
                            use_numbers_as_points=True, score_thresh=0.5):
    """
    Build image/world correspondences using:
      - 'number' detections -> classify to yard values, near/far by y
      - 'hash' detections   -> assign yard by nearest number in x, near/far by y

    Returns:
      image_pts: (N, 2) float32
      field_pts: (N, 2) float32
    """
    h, w = frame_bgr.shape[:2]
    y_mid = h / 2.0

    # 1) Collect and classify all 'number' detections
    number_entries = []
    for det in field_dets:
        if det["label"] != "number":
            continue
        if det["score"] < score_thresh:
            continue

        bbox = det["bbox"]
        u, v = box_center(bbox)

        # side based purely on image y
        side = "near" if v > y_mid else "far"

        yard = classify_yard_from_patch(frame_bgr, bbox, number_model)

        number_entries.append({
            "yard": yard,
            "side": side,
            "u": float(u),
            "v": float(v),
            "bbox": bbox,
        })

    if len(number_entries) == 0:
        raise ValueError("No 'number' detections with sufficient confidence; cannot build homography.")

    # 2) Collect 'hash' detections and associate each with nearest number in x
    image_pts = []
    field_pts = []

    for det in field_dets:
        if det["label"] != "hash":
            continue
        if det["score"] < score_thresh:
            continue

        bbox = det["bbox"]
        u, v = box_center(bbox)

        # find nearest number in x
        # for hashes
        nearest = min(number_entries, key=lambda n: abs(n["u"] - u))
        yard_from_number = nearest["yard"]     # 10, 20, 30, 40, 50, ...
        field_x = float(yard_from_number)      # NO +10
        
        side = "near" if v > y_mid else "far"
        
        image_pts.append([u, v])
        
        if side == "near":
            field_pts.append([field_x, HASH_NEAR_Y])
        else:
            field_pts.append([field_x, HASH_FAR_Y])


    # 3) Optionally add the numbers themselves as extra correspondences
    if use_numbers_as_points:
        for n in number_entries:
            image_pts.append([n["u"], n["v"]])
    
            field_x = float(n["yard"])     # NO +10
    
            if n["side"] == "near":
                field_pts.append([field_x, NUM_NEAR_Y])
            else:
                field_pts.append([field_x, NUM_FAR_Y])


    image_pts = np.asarray(image_pts, dtype=np.float32)
    field_pts = np.asarray(field_pts, dtype=np.float32)

    return image_pts, field_pts
```

```python
def estimate_homography(image_pts, field_pts):
    if len(image_pts) < 4:
        raise ValueError("Need at least 4 correspondences to compute homography")

    H, mask = cv2.findHomography(
        image_pts,
        field_pts,
        method=cv2.RANSAC,
        ransacReprojThreshold=3.0
    )
    return H, mask
```

```python
def project_players_to_field(H, player_dets: List[Dict]):
    players_projected = []

    for det in player_dets:
        label = det["label"]
        bbox  = det["bbox"]
        score = det["score"]

        if score < 0.5:
            continue

        x1, y1, x2, y2 = bbox
        u = (x1 + x2) / 2.0
        v = y2    # bottom of bounding box = feet

        pt = np.array([[[u, v]]], dtype=np.float32)
        proj = cv2.perspectiveTransform(pt, H)[0, 0]   # returns (X, Y)

        players_projected.append({
            "role": label,
            "field_xy": (float(proj[0]), float(proj[1])),
            "score": score
        })

    return players_projected
```

```python
def render_top_down(player_field_points,
                    out_width=1200,
                    out_height=600):
    """
    World coordinates:
      X: 0..100 (goal line to goal line)
      Y: 0..53.3 (sideline to sideline)

    Display coordinates:
      X_display: -10..110 (10-yard endzone on each side)
    """

    display_length = FIELD_LENGTH + 20.0  # 100 + 20 = 120

    canvas = np.zeros((out_height, out_width, 3), dtype=np.uint8)

    sx = out_width / display_length    # yards_display -> pixels
    sy = out_height / FIELD_WIDTH      # yards_y       -> pixels

    def world_to_display_x(X_world: float) -> int:
        """X_world=0 -> display 10, X_world=100 -> display 110."""
        X_display = X_world + 10.0
        return int(X_display * sx)

    # Colors (BGR)
    GREEN_FIELD   = (30, 120, 30)
    WHITE         = (230, 230, 230)
    ENDZONE_LEFT  = (60,  60, 180)
    ENDZONE_RIGHT = (40,  40, 200)
    GRID_TEXT     = (255, 255, 255)
    LOS_COLOR     = (255, 0, 0)   # blue in BGR

    role_radius = {
        "qb": 9,
        "center": 8,
        "skill": 7,
        "lb": 7,
        "db": 7
    }

    role_color = {
        "qb":     (0,   0, 255),
        "skill":  (0, 255,   0),
        "db":     (255, 0,   0),
        "lb":     (0, 255, 255),
        "center": (255, 0, 255)
    }

    # 1. Base field
    canvas[:, :] = GREEN_FIELD

    # 2. Endzones in display coords: [-10,0] and [100,110]
    x_en_l0 = int(0 * sx)         # display -10 -> x=0
    x_en_l1 = int(10 * sx)        # display 0   -> x=10*sx
    x_en_r0 = int(110 * sx)       # display 100 -> x=110*sx
    x_en_r1 = int(120 * sx)       # display 110 -> x=120*sx

    canvas[:, x_en_l0:x_en_l1] = ENDZONE_LEFT
    canvas[:, x_en_r0:x_en_r1] = ENDZONE_RIGHT

    # 3. Sidelines (top/bottom)
    cv2.line(canvas, (0, 0), (out_width - 1, 0), WHITE, 3)
    cv2.line(canvas, (0, out_height - 1), (out_width - 1, out_height - 1), WHITE, 3)

    # 4. Yard lines every 5 yards (world 0..100)
    for x_yard in range(0, 101, 5):
        x_px = world_to_display_x(x_yard)
        thickness = 2 if x_yard % 10 == 0 else 1
        cv2.line(canvas, (x_px, 0), (x_px, out_height - 1), WHITE, thickness)

    # 5. Hash marks
    hash_near_y_px = int(HASH_NEAR_Y * sy)
    hash_far_y_px  = int(HASH_FAR_Y  * sy)

    for x_yard in range(0, 101):
        x_px = world_to_display_x(x_yard)
        cv2.line(canvas, (x_px - 3, hash_near_y_px), (x_px + 3, hash_near_y_px), WHITE, 1)
        cv2.line(canvas, (x_px - 3, hash_far_y_px),  (x_px + 3, hash_far_y_px),  WHITE, 1)

    # 6. Bottom coordinate labels every 10 yards
    axis_y = out_height - 5
    for x_yard in range(0, 101, 10):
        x_px = world_to_display_x(x_yard)
        label = str(x_yard)
        cv2.putText(canvas, label,
                    (x_px - 10, axis_y),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.4,
                    GRID_TEXT,
                    1,
                    cv2.LINE_AA)

    # ---- Compute LOS X (in world coords) robustly ----
    center_x_list = []
    all_x_list = []

    for p in player_field_points:
        Xw, Yw = p["field_xy"]
        all_x_list.append(Xw)
        rnorm = str(p["role"]).upper().strip()
        if rnorm == "C":
            center_x_list.append(Xw)

    if center_x_list:
        center_x_world = float(np.mean(center_x_list))
    elif all_x_list:
        # fallback: median of all players
        center_x_world = float(np.median(all_x_list))
    else:
        center_x_world = None

    # 7. Draw players
    for p in player_field_points:
        Xw, Yw = p["field_xy"]
        role_raw = p["role"]

        rnorm = str(role_raw).upper().strip()
        if rnorm == "QB":
            role = "qb"
        elif rnorm == "C":
            role = "center"
        elif rnorm == "SKILL":
            role = "skill"
        elif rnorm == "LB":
            role = "lb"
        elif rnorm == "DB":
            role = "db"
        else:
            role = rnorm.lower()

        if not (0 <= Xw <= FIELD_LENGTH and 0 <= Yw <= FIELD_WIDTH):
            continue

        x_px = world_to_display_x(Xw)
        y_px = int(Yw * sy)

        r = role_radius.get(role, 6)
        color = role_color.get(role, (255, 255, 255))

        cv2.circle(canvas, (x_px, y_px), r, color, -1)
        cv2.circle(canvas, (x_px, y_px), r, (0, 0, 0), 1)

    # 8. Line of scrimmage (blue dashed) if we found an X
    if center_x_world is not None:
        x_los_px = world_to_display_x(center_x_world)
        y_step = 10
        for y in range(0, out_height, 2 * y_step):
            y1 = y
            y2 = min(y + y_step, out_height - 1)
            cv2.line(canvas, (x_los_px, y1), (x_los_px, y2), LOS_COLOR, 2)

    # 9. Legend (key) in upper left
    legend_x = 20
    legend_y = 30
    line_h   = 20

    legend_entries = [
        ("QB",    "qb"),
        ("Skill", "skill"),
        ("DB",    "db"),
        ("LB",    "lb"),
        ("C",     "center"),
    ]

    for i, (label, key) in enumerate(legend_entries):
        cy = legend_y + i * line_h
        color = role_color.get(key, (255, 255, 255))
        cv2.circle(canvas, (legend_x, cy), 6, color, -1)
        cv2.circle(canvas, (legend_x, cy), 6, (0, 0, 0), 1)
        cv2.putText(canvas, label,
                    (legend_x + 15, cy + 4),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.5,
                    GRID_TEXT,
                    1,
                    cv2.LINE_AA)

    # Legend entry for LOS
    los_label_y = legend_y + len(legend_entries) * line_h
    cv2.line(canvas,
             (legend_x - 5, los_label_y),
             (legend_x + 5, los_label_y),
             LOS_COLOR,
             2)
    cv2.putText(canvas, "Line of scrimmage",
                (legend_x + 15, los_label_y + 4),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                GRID_TEXT,
                1,
                cv2.LINE_AA)

    return canvas
```

```python
def yolo_results_to_dets(results, model):
    """
    Convert Ultralytics YOLO results into:
    [
        {"label": str, "bbox": (x1,y1,x2,y2), "score": float},
        ...
    ]
    """
    boxes_xyxy = results.boxes.xyxy.cpu().numpy()
    cls_ids     = results.boxes.cls.cpu().numpy().astype(int)
    scores      = results.boxes.conf.cpu().numpy()
    names       = model.names

    dets = []
    for box, cid, score in zip(boxes_xyxy, cls_ids, scores):
        x1, y1, x2, y2 = box.tolist()
        dets.append({
            "label": names[cid],
            "bbox": (x1, y1, x2, y2),
            "score": float(score)
        })
    return dets
```

```python
def process_frame(field_results, player_results, field_model, player_model):
    """
    field_results = YOLO output for field model on one frame
    player_results = YOLO output for player model on one frame
    """

    # Convert YOLO outputs to detection dictionaries
    field_dets  = yolo_results_to_dets(field_results,  field_model)
    player_dets = yolo_results_to_dets(player_results, player_model)

    # 1. Build homography correspondences
    img_pts, fld_pts = build_homography_points(field_dets)

    # 2. Estimate homography
    H, mask = estimate_homography(img_pts, fld_pts)

    # 3. Project players into field coordinates
    player_field_points = project_players_to_field(H, player_dets)

    # 4. Produce top-down view
    top_down_img = render_top_down(player_field_points)

    return {
        "H": H,
        "points_image": img_pts,
        "points_field": fld_pts,
        "player_field_points": player_field_points,
        "topdown": top_down_img
    }
```

```python
def classify_yard_from_patch(frame_bgr, bbox, number_model):
    """
    Crop the 'number' bbox from the frame, run the classifier,
    and return an integer yard value (e.g. 10, 20, 30, 40, 50).
    """
    h, w = frame_bgr.shape[:2]
    x1, y1, x2, y2 = map(int, bbox)

    x1 = max(0, min(w-1, x1))
    x2 = max(0, min(w,   x2))
    y1 = max(0, min(h-1, y1))
    y2 = max(0, min(h,   y2))

    crop = frame_bgr[y1:y2, x1:x2]
    if crop.size == 0:
        raise ValueError("Empty crop for number classification")

    cls_results = number_model(crop)[0]

    # Ultralytics classification: probs.top1 is index, names gives label string
    cls_id   = int(cls_results.probs.top1)
    cls_name = cls_results.names[cls_id] if hasattr(cls_results, "names") else number_model.names[cls_id]

    # Expect labels like '10', '20', '30' etc.
    try:
        yard = int(cls_name)
    except ValueError:
        raise ValueError(f"Number classifier returned non-integer label: {cls_name}")

    return yard
```

```python
# ---------- helper for drawing detections ----------
def draw_detections(img, dets, color, thickness=2):
    """
    img: BGR image (in-place)
    dets: list of {"label","bbox","score"}
    """
    for det in dets:
        x1, y1, x2, y2 = map(int, det["bbox"])
        label = det["label"]
        score = det["score"]
        cv2.rectangle(img, (x1, y1), (x2, y2), color, thickness)
        text = f"{label} ({score:.2f})"
        cv2.putText(img, text, (x1, max(0, y1-5)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.4, color, 1, cv2.LINE_AA)

# ---------- 1. Load image ----------
frame_bgr = cv2.imread(img_path)
if frame_bgr is None:
    raise FileNotFoundError(f"Could not read image at {img_path}")

# Ultralytics YOLO works fine with BGR numpy arrays
img_for_yolo = frame_bgr

# ---------- 2. Run YOLO models ----------
field_results  = field_model(img_for_yolo)[0]
player_results = player_model(img_for_yolo)[0]

field_dets  = yolo_results_to_dets(field_results,  field_model)
player_dets = yolo_results_to_dets(player_results, player_model)

# ---------- 3. Visualize detections on original frame ----------
vis_bgr = frame_bgr.copy()
draw_detections(vis_bgr, field_dets,  color=(0, 255, 0))   # green for field
draw_detections(vis_bgr, player_dets, color=(255, 0, 0))   # blue/red for players

vis_rgb = cv2.cvtColor(vis_bgr, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 6))
plt.title("Original frame with YOLO detections")
plt.imshow(vis_rgb)
plt.axis("off")
plt.show()

# ---------- 4. Build homography & project players ----------
img_pts, fld_pts = build_homography_points(field_dets, frame_bgr, number_model)
print("Number of usable field correspondences:", len(img_pts))
if len(img_pts) < 4:
    raise ValueError("Not enough correspondences to estimate homography")

print("Sample image_pts:\n", img_pts[:5])
print("Sample field_pts:\n", fld_pts[:5])

H, mask = estimate_homography(img_pts, fld_pts)
print("Homography matrix H:\n", H)

player_field_points = project_players_to_field(H, player_dets)


# ---------- 5. Render top-down field ----------
topdown_bgr = render_top_down(player_field_points)
topdown_rgb = cv2.cvtColor(topdown_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 4))
plt.title("Top-down field view (projected players)")
plt.imshow(topdown_rgb)
plt.axis("off")
plt.show()
```

```python
import cv2
import numpy as np
from tqdm import tqdm  # optional progress bar, pip install tqdm if needed

def process_video_to_topdown(
    video_path,
    output_path,
    ref_time_sec=5.0,
    conf_thresh=0.5,
    max_frames=None
):
    """
    1) Open input video.
    2) Seek to ref_time_sec, compute homography H_ref from that frame.
    3) From ref_time_sec to end, track players each frame and render top-down.
    4) Save side-by-side (original | top-down) video to output_path.
    """

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        raise FileNotFoundError(f"Could not open video: {video_path}")

    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    ref_frame_idx = int(ref_time_sec * fps)

    print(f"Video FPS: {fps:.2f}, total frames: {total_frames}")
    print(f"Using reference frame index: {ref_frame_idx} (≈ {ref_time_sec} s)")

    # ---------- 1. Grab reference frame ----------
    cap.set(cv2.CAP_PROP_POS_FRAMES, ref_frame_idx)
    ret, ref_frame_bgr = cap.read()
    if not ret:
        cap.release()
        raise RuntimeError(f"Could not read reference frame at index {ref_frame_idx}")

    # Run field model + number model on reference frame
    print("Computing homography from reference frame...")

    field_results_ref = field_model(ref_frame_bgr)[0]
    field_dets_ref    = yolo_results_to_dets(field_results_ref, field_model)

    # Build correspondences using that frame + number_model
    img_pts_ref, fld_pts_ref = build_homography_points(field_dets_ref, ref_frame_bgr, number_model)
    print("Number of usable field correspondences (ref frame):", len(img_pts_ref))

    if len(img_pts_ref) < 4:
        cap.release()
        raise RuntimeError("Not enough correspondences in reference frame to estimate homography")

    H_ref, mask_ref = estimate_homography(img_pts_ref, fld_pts_ref)
    print("Reference homography H:\n", H_ref)

    # ---------- 2. Prepare VideoWriter ----------

    # We'll render top-down once to get its size
    dummy_player_field_points = []
    dummy_topdown = render_top_down(dummy_player_field_points)
    top_h, top_w = dummy_topdown.shape[:2]

    # Ensure original frame has same height as topdown for side-by-side stacking
    ref_h, ref_w = ref_frame_bgr.shape[:2]
    scale = top_h / ref_h
    resized_w = int(ref_w * scale)
    resized_h = top_h

    out_w = resized_w + top_w
    out_h = top_h

    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(output_path, fourcc, fps, (out_w, out_h))

    # ---------- 3. Process frames from ref_frame_idx onward ----------

    # Reset to reference frame index
    cap.set(cv2.CAP_PROP_POS_FRAMES, ref_frame_idx)

    start_frame = ref_frame_idx
    end_frame = total_frames if max_frames is None else min(total_frames, start_frame + max_frames)

    print(f"Processing frames {start_frame} to {end_frame - 1}...")

    for frame_idx in tqdm(range(start_frame, end_frame)):
        ret, frame_bgr = cap.read()
        if not ret:
            break

        # Run player detector on this frame
        player_results = player_model(frame_bgr)[0]
        player_dets    = yolo_results_to_dets(player_results, player_model)

        # Project players using frozen H_ref
        player_field_points = project_players_to_field(H_ref, player_dets)

        # Render top-down
        topdown_bgr = render_top_down(player_field_points)

        # Resize original frame to match topdown height
        frame_resized = cv2.resize(frame_bgr, (resized_w, resized_h), interpolation=cv2.INTER_AREA)

        # Side-by-side concat: [original | topdown]
        combined = np.hstack([frame_resized, topdown_bgr])

        out.write(combined)

    cap.release()
    out.release()
    print(f"Saved output video to: {output_path}")
```

```python
import os
import glob
import numpy as np
import cv2

def evaluate_geometry_first_100(folder_path, image_extensions=(".png", ".jpg", ".jpeg")):
    """
    Processes the FIRST 100 images in folder_path.
    For each image:
      - Compute H
      - Project players
      - Compute center-hash and QB-center geometry errors

    Prints final aggregated metrics ONLY.
    """
    # Collect and limit paths
    paths = []
    for ext in image_extensions:
        paths.extend(glob.glob(os.path.join(folder_path, f"*{ext}")))
    paths = sorted(paths)[:100]   # <<< only first 100 images

    if len(paths) == 0:
        print("No images found.")
        return None

    center_hash_errors = []
    qb_center_errors = []

    for img_path in paths:
        frame_bgr = cv2.imread(img_path)
        if frame_bgr is None:
            continue

        try:
            # Field detections
            field_results = field_model(frame_bgr)[0]
            field_dets    = yolo_results_to_dets(field_results, field_model)

            # Homography correspondences
            img_pts, fld_pts = build_homography_points(field_dets, frame_bgr, number_model)
            if img_pts.shape[0] < 4:
                continue

            # Fit homography
            H, mask = estimate_homography(img_pts, fld_pts)
            if H is None:
                continue

            # Player detections
            player_results = player_model(frame_bgr)[0]
            player_dets    = yolo_results_to_dets(player_results, player_model)

            # Project to field coords
            player_field_points = project_players_to_field(H, player_dets)

            # Geometry error
            geom = evaluate_play_geometry(player_field_points)
            if geom is None:
                continue

            center_hash_errors.append(geom["center_hash_error_yards"])
            qb_center_errors.append(geom["qb_center_alignment_error_yards"])

        except Exception:
            continue

    # Aggregate
    if not center_hash_errors:
        print("No valid frames contained both CENTER and QB with a valid homography.")
        return None

    center_hash_errors = np.array(center_hash_errors, dtype=float)
    qb_center_errors   = np.array(qb_center_errors, dtype=float)

    metrics = {
        "frames_used": len(center_hash_errors),
        "center_hash_mean": float(center_hash_errors.mean()),
        "center_hash_median": float(np.median(center_hash_errors)),
        "center_hash_max": float(center_hash_errors.max()),
        "qb_center_mean": float(qb_center_errors.mean()),
        "qb_center_median": float(np.median(qb_center_errors)),
        "qb_center_max": float(qb_center_errors.max()),
    }

    # Final print only
    print("\n=== Geometry Evaluation Over First 100 Images ===")
    print(f"Frames used: {metrics['frames_used']}")
    print(f"Center–Hash Error (yards): "
          f"mean={metrics['center_hash_mean']:.3f}, "
          f"median={metrics['center_hash_median']:.3f}, "
          f"max={metrics['center_hash_max']:.3f}")
    print(f"QB–Center Alignment (yards): "
          f"mean={metrics['qb_center_mean']:.3f}, "
          f"median={metrics['qb_center_median']:.3f}, "
          f"max={metrics['qb_center_max']:.3f}")

    return metrics
```
